```python
#Data Preprocessing
```

```python
#Data Cleaning
```

```python
#converting date column to datetime
df["Date"]=pd.to_datetime(df["Date"])
df.sort_values("Date",inplace=True)
df.reset_index(drop=True,inplace=True)
df
```

```python
#treating missing values
```

```python
#fill missing values forward
df.fillna(method='ffill',inplace=True)
#optionally dropping if still any missing values at top(after fill)
df.dropna(inplace=True)
df
```

```python
df.isnull().sum()
```

```python
#check for zero or invalid values 
```

```python
print((df[['Open', 'High', 'Low', 'Close','Volume']]==0).sum())
```

```python
#replace 0 and Nan with forward fill
df[['Open', 'High', 'Low', 'Close','Volume']]=df[['Open', 'High', 'Low', 'Close','Volume']].replace(0,np.nan)
df.fillna(method='ffill',inplace=True)
df
```

```python
df.isnull().sum()
```

```python
# Feature Engineering
```

```python
df['Close_Lag1'] = df['Close'].shift(1)
df['MA7'] = df['Close'].rolling(window=7).mean()
df['MA21'] = df['Close'].rolling(window=21).mean()
df['Pct_Change'] = (df['Close'] - df['Open']) / df['Open']
df['Daily_Return'] = df['Close'].pct_change()
df['Volatility'] = df['Daily_Return'].rolling(window=7).std()
df['DayOfWeek'] = df['Date'].dt.dayofweek  # 0=Monday, 6=Sunday

#drop date column
df.drop(columns=['Date'],inplace=True)
# Drop NaNs from rolling & lag features
df.dropna(inplace=True)

# Final feature set
#features = [
    #'Open', 'High', 'Low', 'Volume',
    #'Close_Lag1', 'MA7', 'MA21',
    #'Pct_Change', 'Volatility', 'DayOfWeek'
#]
df
```

```python
df.columns.tolist()
```

```python
col=['Open','High','Low','Close','Adj Close','Volume','Close_Lag1','MA7','MA21','Pct_Change','Daily_Return','Volatility']
```

```python
#checking outliers
for i in col:
    print(sns.boxplot(df[i]))
    plt.title("Checking outliers")
    plt.xlabel(i)
    plt.grid(True)
    plt.show()
```

```python
for i in col:
    Q1=df[i].quantile(0.25)
    Q3=df[i].quantile(0.75)
    IQR=Q3-Q1
    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    df[i]=np.where(df[i]>upper,upper,np.where(df[i]<lower,lower,df[i]))
    
```

```python
for i in col:
    print(sns.boxplot(df[i]))
    plt.title("Checking outliers")
    plt.xlabel(i)
    plt.grid(True)
    plt.show()
```

```python
for i in col:
    print(df[i].skew())

#Data transformation
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
df[col]=scaler.fit_transform(df[col])
df[col]
```
